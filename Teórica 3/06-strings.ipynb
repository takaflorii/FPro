# 6 Data types: Strings

## 6.1 A compound data type

- So far we have seen built-in types like `int`, `float`, `bool`,
    `str` (and very briefly lists and pairs).

- Strings, lists, and pairs are qualitatively different from the
    others because they are made up of smaller pieces.

- Types that comprise smaller pieces are called **collections** or
    **compound data types**.

- For instance, strings are made up of smaller strings each
    containing one **character** in a particular order from left to
    right.

- Depending on what we are doing, we may want to treat a value of a compound data type as a single thing.



### Text sequence type — `str`

  - Textual data in Python is handled with `str` objects, or strings.
  
  - **Strings are immutable** sequences of Unicode code points.

  - Being **immutable** means that once a string is created, you cannot change its contents. Any operation that seems to modify a string will actually create a new string instead of changing the original one.
  
  

- String literals are written in a variety of ways:

    - Single quotes: `'allows embedded "double" quotes'`;

    - Double quotes: `"allows embedded 'single' quotes"`;

    - Triple quoted: `'''Three single quotes'''`, `"""Three double quotes"""`.

  - Triple quoted strings may span multiple lines - all associated whitespace will be included in the string literal (used in *docstrings* and comments).

  - A string that contains no characters, e.g. `""`, often referred to as the **empty string**, is still considered to be a string.

    - this is not the same as a non-empty string `" "` that actually contains a character that happens to be a white space.

## 6.5 Traversal and the `for` loop

- A lot of computations involve processing a string *one character at
    a time*.

- These often start at the beginning, select each character in turn, do
    something with it, and continue until the end.

- This pattern of processing is called a **traversal**.

  ```python
  word = "Banana"
  for letter in word:
    print(letter)
  ```

<https://github.com/asfmendes/FPro-public/tree/master/lectures/06/straversal.py>

Since a string is simply a sequence of characters, the `for` loop *iterates* over each character automatically.

````python
words = "Snickerdoodle"
for l in words:
    print(l)
````

Recall that the loop variable takes on each value in the sequence of names. The body is performed once for each name.

````python
for aname in ["Eric", "Graham", "John", "Michael", "Terry"]:
  for char in aname:
    invitation = "Hi " + char + ".  Please come to my party on Saturday!"
    print(invitation)
  print("Prox")
````

### Traversal and looping: by index

- It is also possible to use the `range()` function to systematically generate the indices of a string.

- The `for` loop can then be used to iterate over these positions.

- These positions can be used together with the indexing operator to access the individual characters in the string.


Try this (and then see it using [Python Tutor](http://www.pythontutor.com/visualize.html#mode=edit)):

````python
fruit = "apple"
for idx in range(len(fruit)): # if you used 5 instead of len(fruit), 
                              # this code would not work if you changed 
                              # to a string of a different size.
    current_char = fruit[idx]
    print(current_char)
````

Likewise:


````python
fruit = "apple"

position = 0
while position < len(fruit):
    print(fruit[position])
    position = position + 1
````

Having the indices is useful when you need to access more than a single character in the loop.

````python
fruit = "gooseberry"
for idx in range(len(fruit)):
    # why the disjunction?
    if idx == len(fruit) - 1 or fruit[idx] != fruit[idx+1]:  
      print(fruit[idx])
````

- Again, this also applies to other sequence types (e.g., tuples and lists).

## 6.10 The `enumerate()` function



- `enumerate()` is a built-in function of Python.

  ```python
  enumerate(iterable, start=0)
  ```

- `enumerate()` pairs each element of the sequence with its order count.

- Returns an *enumerate object*.

- The `iterable` argument must be a sequence (e.g., a string), an *iterator*, or some other object which supports iteration.

We can consume the result of `enumerate` with a `for` statement:

````python
quote = "Always look"

print("Pairs:")
for pair in enumerate(quote):
    print(pair)
````

````python
print("\nCounter Char:")
for counter, char in enumerate(quote):
    print(counter, char)
````

## 6.11 The built-in methods `find()` and `count()`

#### `find()`
What does the following function do?

````python
def my_find(haystack, needle):
  for index, letter in enumerate(haystack):
      if letter == needle:
          return index  
  return -1
````

Compare the output of the code above with what Python does itself:

````python
haystack = "Bananarama!"
print("Mine:", my_find(haystack,'a'))
print("Python's:", haystack.find('a', 7))
````



- The `find()` method searches for a substring inside a string and returns the index of its first occurrence.

- It can find substrings, not just single characters. 

  ```python
    >>> "banana".find("nan")
    2
  ```
- You can provide optional second and third arguments to limit the search to a specific slice of the string, as in `string.find(substring, start, end)`

  ```python
    >>> "banana".find("na", 3)
    4
  ```

  <br /> 
| Parameter   | Meaning                                                         |
| ----------- | --------------------------------------------------------------- |
| `substring` | The string you are searching for                                |
| `start`     | Optional. The index to start searching from (default `0`)       |
| `end`       | Optional. The index to stop searching (default `end of string`) |

<br />

## 6.13 Optional parameters

- The functions we've defined so far have a fixed number of parameters, but sometimes with want a little more flexibility.

- Functions can have **optional parameters** as long as we define the *default value* to be used when that argument is not provided in a call.

- Let's say we want to find the locations of the occurrence of a character in a string starting at a given position.

- We can modify the `find()` function as a `find_2()` function, adding a third parameter for the starting position in the search string.

- Better still, we can combine `find()` and `find_2()` using an optional parameter, if we assume 0 as the default starting position:

  ```python
    def find_2(haystack, needle, start=0):
        for index,letter in enumerate(haystack[start:]):
            if letter == needle:
                return index + start
        return -1
  ```

<https://github.com/asfmendes/FPro-public/tree/master/lectures/06/optional.py>

Let's do it:

````python
def find_2(haystack, needle, start=0):
    for index,letter in enumerate(haystack[start:]):
        if letter == needle:
            return index + start
    return -1

print(find_2("banana", "a", 2) == 3)
print(find_2("banana", "a") == 3)
````

Another way of doing it with a `while` loop:

````python
def find_3(astring, needle, start=0):
    """
    Find and return the index of a char (the needle) in a string (astring).
    Return -1 if the needle does not occur in astring.
    """
    ix = start
    while ix < len(astring):
        if astring[ix] == needle:
            return ix
        else:
            ix = ix + 1
    return -1

print(find_3('banana', 'a', 2))
````

## 6.16 String method `format()`

- The easiest and most powerful way to format a string in Python3 is
    to use the `format()` method.

- The template string contains place holders, `"... {0} ... {1} ... {2} ..."`.

- The `format()` method substitutes its arguments into the place holders.

- To see how this works, let's start with a few examples:

  ```python
    phrase = "His name is {0}!".format("Arthur")
    print(phrase)

  ```

<https://github.com/asfmendes/FPro-public/tree/master/lectures/06/format.py>

Try it here:

````python
name = "Alice"
age = 10
phrase = "I am {0} and I am {1} years old.".format(name, age)
print(phrase)
````

````python
phrase = "I am {1} and I am {0} years old.".format(age, name)
print(phrase)
````

You can omit the position identifiers in the placeholders, and the argument order will be used by default:

````python
name = "Alice"
age = 10
phrase = "I am {} and I am {} years old.".format(name, age)
print(phrase)
````

You can also repeat the identifiers:

````python
name = "James"
surname = "Bond"
phrase = "My name is {1}. {0} {1}.".format(name, surname)
print(phrase)
````

### Format specification

- Each of the replacement fields can also contain a **format specification**.

- This modifies how the substitutions are made into the template, and can control things like:

    - Whether the field is aligned to the left `<`, center `^`, or right `>`;

    - The width allocated to the field within the result string (a number like 10);

    - The type of conversion (we'll initially only force conversion to float (`f`) or perhaps we'll ask integer numbers to be converted to hexadecimal using `x`);

    - If the type conversion is a float, you can also specify how many decimal places are wanted (such as `.2f` to work with currencies with two decimal places).

<https://github.com/asfmendes/FPro-public/tree/master/lectures/06/formatspec.py>

See the differences:

````python
cost = 10.10
payment = 9.60
change = cost - payment
print("{:05.2f}€".format(change)) # placeholder values formated as, float, 
                                  # 5 characters wide, padded with zeros, 
                                  # 2 decimal places.
print("{:015.3f}€".format(change)) # placeholder values formated as, float, 
                                   # 15 characters wide, padded with zeros, 
                                   # 3 decimal places.
````

| Format Spec | Meaning                                 | Example                                  |
| ----------- | --------------------------------------- | ---------------------------------------- |
| `d`         | Integer (decimal)                       | `"{:d}".format(42)` → `42`               |
| `f`         | Floating-point                          | `"{:.2f}".format(3.14159)` → `3.14`      |
| `.2f`       | Float with 2 decimal places             | `"{:.2f}".format(2.5)` → `2.50`          |
| `05d`       | Integer, width 5, zero-padded           | `"{:05d}".format(42)` → `00042`          |
| `08.2f`     | Float, width 8, 2 decimals, zero-padded | `"{:08.2f}".format(12.3)` → `00012.30`   |
| `<`         | Left-align in given width               | `"{:<10}".format("hi")` → `'hi        '` |
| `>`         | Right-align in given width              | `"{:>10}".format("hi")` → `'        hi'` |
| `^`         | Center-align in given width             | `"{:^10}".format("hi")` → `'    hi    '` |
| `%`         | Percentage                              | `"{:.1%}".format(0.1234)` → `12.3%`      |
| `,`         | Thousands separator                     | `"{:,}".format(1234567)` → `1,234,567`   |
| `b`         | Binary                                  | `"{:b}".format(5)` → `101`               |
| `x`         | Hexadecimal (lowercase)                 | `"{:x}".format(255)` → `ff`              |
| `X`         | Hexadecimal (uppercase)                 | `"{:X}".format(255)` → `FF`              |


## 6.17 Formatted string literals

From Python 3.6 onwards there are also *formatted string literals* (or *f-strings*), an alternative way to format strings:

  ```python
  name = "Eric"
  age = 74
  f"Hello, {name}. You are {age}."
  ```

Because f-strings are evaluated at runtime, you can put any and all valid Python expressions in them:

````python
f"{2 * 37}"
````

But you could also call functions. Here’s an example:

````python
def to_lowercase(input):
    return input.lower()

name = "Eric Idle"
f"{to_lowercase(name)} is funny."
````

You also have the option of calling a method directly:

````python
f"{name.lower()} is funny."
````

And you can still use the formatting options that we've seen above:

````python
cost = 10.10
payment = 9.60
print(f"{cost - payment:05.2f}€".format(change))
````
